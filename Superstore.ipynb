```python
df=pd.read_csv("Sample - Superstore.csv",encoding='latin-1')
```

```python
print("Rows and columns:", df.shape)
print("Column names:", df.columns)
print(df.info())
```

```python
print("Missing values:")
print(df.isnull().sum())

print("Duplicate records:", df.duplicated().sum())
```

```python
print("Total sales:", df["Sales"].sum())
print("Total profit:", df["Profit"].sum())
print("Average order-line sales:", df["Sales"].mean())
```

```python
print("Sales by category:")
print(df.groupby("Category")["Sales"].sum().sort_values(ascending=False))
```

```python
print("Profit by sub-category:")
print(df.groupby("Sub-Category")["Profit"].sum().sort_values())
```

```python
print("Sales by region:")
print(df.groupby("Region")["Sales"].sum().sort_values(ascending=False))
```

```python
print("Sales by segment:")
print(df.groupby("Segment")["Sales"].sum().sort_values(ascending=False))
```

```python
print("Discount and average profit:")
print(df.groupby("Discount")["Profit"].mean())
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Month"] = df["Order Date"].dt.to_period("M")
print("Monthly sales:")
print(df.groupby("Month")["Sales"].sum())
```
