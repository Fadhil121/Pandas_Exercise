```python
import pandas as pd
```

```python
satisfaction_scores=pd.Series([8,6,9,4,7,10,5,8])
print(satisfaction_scores)
```

```python
satisfaction_scores=pd.Series([8,6,9,4,7,10,5,8],index=["C001","C002","C003","C004","C005","C006","C007","C008"])
print(satisfaction_scores)
```

```python
satisfaction_scores=pd.Series({"C001":8,"C002":6,"C003":9,"C004":4,"C005":7,"C006":10,"C007":5,"C008":8})
print(satisfaction_scores)
```

```python
print("Values :",satisfaction_scores.values)
```

```python
print(satisfaction_scores.index)
```

```python
print("Data Type :",satisfaction_scores.dtype)
```

```python
satisfaction_scores.name='Customer Satisfaction'
print(satisfaction_scores)
```

```python
print("Score of Customer:",satisfaction_scores.loc["C003"])
```

```python
print("Score of 6th customer:",satisfaction_scores.iloc[5])
```

```python
print("Scores of customers:",satisfaction_scores.loc[["C001","C003","C006","C008"]])
```

```python
print("Scores of customers:",satisfaction_scores.iloc[[0,2,5,7]])
```

```python
print("Scores of customers:",satisfaction_scores.loc[["C002","C007"]])
```

```python
print("Total Number of Customers:",satisfaction_scores.count())
```

```python
print("Total Satisfaction Scores:",satisfaction_scores.sum())
```

```python
print("Average Satisfaction Scores:",satisfaction_scores.mean())
```

```python
print("Highest Satisfaction Scores:",satisfaction_scores.max())
```

```python
print("Lowest Satisfaction Scores:",satisfaction_scores.min())
```

```python
print("Median Satisfaction Scores:",satisfaction_scores.median())
```
