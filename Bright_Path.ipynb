```python
import pandas as pd
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
revenue_values  = [42, 45, 50, 48, 55, 60, 58, 62, 65, 70, 75, 85]
expense_values  = [38, 40, 42, 47, 50, 52, 55, 60, 61, 64, 66, 72]
```

```python
revenue=pd.Series(revenue_values,index=months,name="revenue")
expense=pd.Series(expense_values,index=months,name="expense")
print(revenue)
print(expense)
```

```python
a = [42, 45, 50]
b = [38, 40, 42]
print(a + b)
```

```python
profit_list = []

for i in range(12):
    profit_list.append(revenue.iloc[i] - expense.iloc[i])

print("Profit using loop:", profit_list)
```

```python
profit_series = revenue - expense
profit_list_clean = profit_series.tolist()
print("Clean profit list:", profit_list_clean)
```

```python
profit=revenue-expense
profit.name="profit"
print(f"monthly profit \n {profit}")
```

```python
profit = revenue - expense
profit.name = "Profit"
print(profit)
```

```python
margin=round(profit/revenue*100,1)
margin.name="profit margin %"
print(f"margin\n {margin}")
```

```python
total_revenue = revenue.sum()
total_expenses = expense.sum()
total_profit = profit.sum()
overall_margin = total_profit / total_revenue * 100
print("Total revenue:", total_revenue)
print("Total expenses:", total_expenses)
print("Total profit:", total_profit)
print("Overall margin:", round(overall_margin, 1), "%")
```

```python
q1 = profit.loc["Jan":"Mar"].sum()
q2 = profit.loc["Apr":"Jun"].sum()
q3 = profit.loc["Jul":"Sep"].sum()
q4 = profit.loc["Oct":"Dec"].sum()

quarters = pd.Series([q1, q2, q3, q4],
                     index=["Q1", "Q2", "Q3", "Q4"])

print("Quarterly profit:")
print(quarters)
print("Strongest quarter:", quarters.idxmax())
```

```python
average = profit.mean()

print("Average monthly profit:", average)
print("Months above average:")
print(profit[profit > average])
```

```python
p=lambda n: "Strong" if (n>=8) else "Stable" if(n<8)&(n>=4) else "Weak" if (n<4) else "fail"
profit_status=profit.apply(p)
print(f'profit_status \n {profit_status}')
```

```python
status_counts=profit_status.value_counts()
print(f"number of status {status_counts}")
status_pct=profit_status.value_counts(normalize=True)*100
print(f"percentage of months in each status\n{status_pct}")
```
