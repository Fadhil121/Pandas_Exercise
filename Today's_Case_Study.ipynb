```python
import pandas as pd
```

```python
#Create a series score from the list of scores with student names as the index
student=pd.Series([78,85,92,67,74,88,95,61], index=["Anu","Binu","Cathy","Deepa","Ebin","Fathima","Gopal","Hari"])
print(student)
```

```python
#Create a series dept from a Python dictionary mapping student to department
m={"Anu":"CS","Binu":"IT","Cathy":"CS","Deepa":"BCA","Ebin":"IT","Fathima":"CS","Gopal":"CS","Hari":"BCA"}
w=pd.Series(m)
print(w)
```

```python
#Set the name of scores to "Test Score" and display it.
score=pd.Series([78,85,92,67,74,88,95,61])
score.name="Test Score"
print(score)
```
