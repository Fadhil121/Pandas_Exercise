```python
enrol_list = [
    "Python", "SQL", "Excel", "Python", "Power BI", "SQL", "Python",
    "Excel", "Tableau", "Python", "SQL", "Python", "Power BI", "Excel",
    "Python", "SQL", "Statistics", "Python", "Excel", "SQL", "Power BI",
    "Python", "SQL", "Excel", "Python", "Tableau", "SQL", "Power BI",
    "Python", "Excel", "Python", "SQL", "Power BI", "Statistics", "Python",
    "Tableau", "SQL", "Excel", "Machine Learning", "Power BI"
]

mode_list = [
    "Online", "Classroom", "Online", "Online", "Online", "Online", "Hybrid",
    "Online", "Online", "Online", "Classroom", "Online", "Hybrid", "Online",
    "Classroom", "Online", "Hybrid", "Hybrid", "Hybrid", "Hybrid", "Online",
    "Hybrid", "Online", "Hybrid", "Classroom", "Online", "Online", "Online",
    "Classroom", "Online", "Hybrid", "Classroom", "Hybrid", "Online", "Classroom",
    "Online", "Online", "Hybrid", "Online", "Online"
]

rating_list = [5, 4, 4, 3, 5, 2, 4, 5, 3, 4, 1, 5, 4, 3, 4, 5, 2, 4, 5, 4, 3, 4, 5, 2, 4]
```

```python
enrolments = pd.Series(enrol_list)
enrolments.name="Course Enrolled"
print(enrolments)
```

```python
mode=pd.Series(mode_list)
mode.name="Study Mode"
print(mode)
```

```python
print(mode.value_counts())
percent = (mode.value_counts(normalize=True)*100)
print("as percentage =",percent)
```

```python
print("most perferred mode=\n",percent.idxmax())
print("combined percentage \n",percent.loc["Online"]+percent.loc["Hybrid"])
```

```python
ratings=pd.Series(rating_list)
ratings.name="Ratings"
print(ratings)
```

```python
print("Number:", ratings.count())
print("Average:", round(ratings.mean(), 2))
print("Highest:", ratings.max())
print("Lowest:", ratings.min())
```

```python
print(ratings.value_counts(ascending=False))
```

```python
sentiment= lambda m: "postive" if m>=4 else "neutral" if m==3 else "negative"
sentiments= ratings.apply(sentiment)
print(sentiments)
sentiments.index=ratings
print(sentiments)
```

```python
sentiment_share = (sentiments.value_counts(normalize=True)*100).round(1)
print(sentiment_share)
```

```python
def portfolio_action(pct):
    if pct >= 20:
        return "Expand"
    elif pct >= 10:
        return "Maintain"
    else:
        return "Review"

actions = counts.apply(portfolio_action)
print(actions)
```
